## NUMPY

```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
print(arr)
```

```python
arr2d = np.array([[1, 2, 3], [4, 5, 6]])
print(arr2d.shape)
reshaped = arr2d.reshape(3, 2)
print(reshaped)
```

## SCIKIT-LEARN

### 1. Train-test split
Divide data for training and testing.

```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
print(X_train.shape, X_test.shape)
```

### 2. Decision Tree classifier
Train classifier and make prediction.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)
clf = DecisionTreeClassifier()
clf.fit(X, y)
pred = clf.predict([[5.1, 3.5, 1.4, 0.2]])
print(pred)
```

### 3. Scaling features
Normalize features to 0 mean and unit variance.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled[:2])
```

### 4. Model accuracy
Evaluate model performance on test data.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
clf = DecisionTreeClassifier()
clf.fit(X_train, y_train)
accuracy = clf.score(X_test, y_test)
print(f"Accuracy: {accuracy}")
```

### 5. Cross-validation
Test model across multiple data splits.

```python
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)
clf = DecisionTreeClassifier()
scores = cross_val_score(clf, X, y, cv=5)
print(f"CV scores: {scores}, Mean: {scores.mean()}")
```

## KERAS

### 1. Sequential model
Create simple neural network architecture.

```python
import keras
from keras.models import Sequential
from keras.layers import Dense
model = Sequential([
    Dense(10, activation='relu', input_dim=4),
    Dense(3, activation='softmax')
])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
print(model.summary())
```

### 2. Training model
Train network on data.

```python
import keras
from keras.models import Sequential
from keras.layers import Dense
from keras.datasets import mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()
X_train = X_train.reshape(-1, 784).astype('float32') / 255
X_test = X_test.reshape(-1, 784).astype('float32') / 255
model = Sequential([Dense(128, activation='relu', input_dim=784), Dense(10, activation='softmax')])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=5, batch_size=32)
```

### 3. Model prediction
Predict on new data.

```python
predictions = model.predict(X_test[:5])
print(predictions)
```

### 4. Layers and activation
Build layer-by-layer, showing different activations.

```python
from keras.models import Sequential
from keras.layers import Dense
model = Sequential()
model.add(Dense(64, activation='relu', input_dim=10))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
print(model.summary())
```

### 5. Evaluation
Test trained model on test set.

```python
loss, accuracy = model.evaluate(X_test, y_test)
print(f"Loss: {loss}, Accuracy: {accuracy}")
```
