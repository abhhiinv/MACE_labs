```python
# M24CA1L306 Data Science Lab - Lab Exercise #03
# Student: Abhinav | MAC25MCA-2002 | S3 MCA (2025-27)
# Dataset: iris.csv

import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt

matplotlib.use('Agg')
import seaborn as sns
from pandas.plotting import parallel_coordinates, andrews_curves

# Load dataset
df = pd.read_csv("Iris.csv")
df.drop(columns=["Id"], inplace=True)  # Id column is not a feature

print(df.head())
print(df.describe())
print(df["Species"].value_counts())

features = ["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]
species_list = df["Species"].unique()
```

```python
# ─────────────────────────────────────────────
# 1. Histogram
# ─────────────────────────────────────────────
fig, axes = plt.subplots(2, 2, figsize=(10, 7))
for ax, feature in zip(axes.flatten(), features):
    for sp in species_list:
        ax.hist(df[df["Species"] == sp][feature], alpha=0.6, label=sp, bins=15)
    ax.set_title(f"Histogram - {feature}")
    ax.set_xlabel(feature)
    ax.set_ylabel("Frequency")
    ax.legend(fontsize=7)
plt.suptitle("Histogram of Features by Species")
plt.tight_layout()
plt.savefig("lab03_01_histogram.png", dpi=100)
plt.close()

# Findings: Petal features show clear separation between species,
# especially Setosa which clusters at much lower values.
# Sepal features overlap more across species.
```

```python
# ─────────────────────────────────────────────
# 2. Quartile (Box) Plot
# ─────────────────────────────────────────────
fig, axes = plt.subplots(2, 2, figsize=(10, 7))
for ax, feature in zip(axes.flatten(), features):
    df.boxplot(column=feature, by="Species", ax=ax)
    ax.set_title(feature)
    ax.set_xlabel("Species")
    ax.set_ylabel(feature)
plt.suptitle("Box Plots of Features by Species")
plt.tight_layout()
plt.savefig("lab03_02_boxplot.png", dpi=100)
plt.close()

# Findings: Setosa has the smallest petal length and width with very low
# spread. Virginica has the largest values overall. Sepal width shows
# more overlap than petal features.
```

```python
# ─────────────────────────────────────────────
# 3. Distribution Chart (KDE)
# ─────────────────────────────────────────────
fig, axes = plt.subplots(2, 2, figsize=(10, 7))
for ax, feature in zip(axes.flatten(), features):
    for sp in species_list:
        subset = df[df["Species"] == sp][feature]
        subset.plot.kde(ax=ax, label=sp)
    ax.set_title(f"Distribution - {feature}")
    ax.set_xlabel(feature)
    ax.legend(fontsize=7)
plt.suptitle("KDE Distribution of Features by Species")
plt.tight_layout()
plt.savefig("lab03_03_distribution.png", dpi=100)
plt.close()

# Findings: KDE curves for petal length and width show three clearly
# separated peaks - one per species. Sepal distributions overlap
# heavily, confirming they are weaker classifiers.
```

```python
# ─────────────────────────────────────────────
# 4. Scatter Plot - Petal Length vs Petal Width
# ─────────────────────────────────────────────
plt.figure(figsize=(7, 5))
colors = {"Iris-setosa": "red", "Iris-versicolor": "blue", "Iris-virginica": "green"}
for sp in species_list:
    subset = df[df["Species"] == sp]
    plt.scatter(subset["PetalLengthCm"], subset["PetalWidthCm"],
                label=sp, alpha=0.7, color=colors[sp])
plt.title("Scatter Plot - Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.legend()
plt.tight_layout()
plt.savefig("lab03_04_scatter.png", dpi=100)
plt.close()

# Findings: The three species form well-separated clusters on petal
# dimensions. Setosa occupies the bottom-left corner (small petals),
# while Virginica has the largest petals. Some overlap exists between
# Versicolor and Virginica.
```

```python
# ─────────────────────────────────────────────
# 5. Scatter Multiple - Sepal vs Petal features
# ─────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for sp in species_list:
    subset = df[df["Species"] == sp]
    axes[0].scatter(subset["SepalLengthCm"], subset["SepalWidthCm"],
                    label=sp, alpha=0.7, color=colors[sp])
    axes[1].scatter(subset["PetalLengthCm"], subset["PetalWidthCm"],
                    label=sp, alpha=0.7, color=colors[sp])
axes[0].set_title("Sepal Length vs Sepal Width")
axes[0].set_xlabel("Sepal Length (cm)")
axes[0].set_ylabel("Sepal Width (cm)")
axes[0].legend(fontsize=7)
axes[1].set_title("Petal Length vs Petal Width")
axes[1].set_xlabel("Petal Length (cm)")
axes[1].set_ylabel("Petal Width (cm)")
axes[1].legend(fontsize=7)
plt.suptitle("Scatter Multiple - Sepal vs Petal Dimensions")
plt.tight_layout()
plt.savefig("lab03_05_scatter_multiple.png", dpi=100)
plt.close()

# Findings: Sepal scatter shows significant overlap between Versicolor
# and Virginica. The petal scatter shows much cleaner separation,
# confirming petal dimensions are better discriminators.
```

```python
# ─────────────────────────────────────────────
# 6. Scatter Matrix (Pair Plot)
# ─────────────────────────────────────────────
palette = {
    "Iris-setosa": "red",
    "Iris-versicolor": "blue",
    "Iris-virginica": "green"
}
pair = sns.pairplot(df, hue="Species", palette=palette, diag_kind="kde")
pair.fig.suptitle("Scatter Matrix (Pair Plot)", y=1.02)
pair.fig.savefig("lab03_06_scatter_matrix.png", dpi=100)
plt.close()

# Findings: Petal length and petal width pair shows the cleanest
# separation across all species. Diagonal KDE plots confirm distinct
# distributions for Setosa vs. the other two species.
```

```python
# ─────────────────────────────────────────────
# 7. Bubble Chart - Sepal dimensions, bubble size = Petal Width
# ─────────────────────────────────────────────
plt.figure(figsize=(8, 6))
for sp in species_list:
    subset = df[df["Species"] == sp]
    plt.scatter(
        subset["SepalLengthCm"],
        subset["SepalWidthCm"],
        s=subset["PetalWidthCm"] * 100,
        alpha=0.5,
        label=sp,
        color=colors[sp]
    )
plt.title("Bubble Chart - Sepal Dimensions (bubble size = Petal Width)")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.legend()
plt.tight_layout()
plt.savefig("lab03_07_bubble.png", dpi=100)
plt.close()

# Findings: Virginica samples have the largest bubbles (widest petals)
# and tend toward higher sepal lengths. Setosa has the smallest bubbles
# despite being clustered at higher sepal widths.
```

```python
# ─────────────────────────────────────────────
# 8. Density Chart (2D KDE) - Petal Length vs Petal Width
# ─────────────────────────────────────────────
plt.figure(figsize=(8, 6))
for sp, color in colors.items():
    subset = df[df["Species"] == sp]
    sns.kdeplot(
        x=subset["PetalLengthCm"],
        y=subset["PetalWidthCm"],
        fill=True,
        alpha=0.4,
        label=sp,
        color=color
    )
plt.title("2D Density Chart - Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.legend()
plt.tight_layout()
plt.savefig("lab03_08_density.png", dpi=100)
plt.close()

# Findings: The three density regions are largely non-overlapping on
# petal dimensions. Setosa is isolated at the lower-left. Versicolor
# and Virginica density regions are adjacent but distinguishable.

```

```python
# ─────────────────────────────────────────────
# 9. Parallel Coordinates Chart
# ─────────────────────────────────────────────
plt.figure(figsize=(10, 5))
parallel_coordinates(df, "Species", color=["red", "blue", "green"], alpha=0.4)
plt.title("Parallel Coordinates Chart")
plt.ylabel("Feature Value (cm)")
plt.tight_layout()
plt.savefig("lab03_09_parallel.png", dpi=100)
plt.close()

# Findings: Setosa lines stay consistently low on petal features while
# varying more on sepal width. Virginica lines are high across petal
# features. Lines for Versicolor and Virginica cross frequently on
# sepal dimensions, confirming their overlap there.
```

```python
# ─────────────────────────────────────────────
# 10. Deviation Chart - Mean deviation from overall mean per species
# ─────────────────────────────────────────────
overall_mean = df[features].mean()
species_means = df.groupby("Species")[features].mean()
deviation = species_means - overall_mean

deviation.T.plot(kind="bar", figsize=(10, 5), color=["red", "blue", "green"], alpha=0.8)
plt.axhline(0, color="black", linewidth=0.8, linestyle="--")
plt.title("Deviation Chart - Species Mean vs Overall Mean")
plt.xlabel("Feature")
plt.ylabel("Deviation from Overall Mean (cm)")
plt.xticks(rotation=15)
plt.legend(title="Species",loc="upper left")
plt.tight_layout()
plt.savefig("lab03_10_deviation.png", dpi=100)
plt.close()

# Findings: Setosa deviates negatively on petal features (smaller than
# average) and positively on sepal width. Virginica deviates positively
# on all petal features (larger than average). Versicolor stays close
# to the overall mean across most features.
```

```python
# ─────────────────────────────────────────────
# 11. Andrews Curves
# ─────────────────────────────────────────────
plt.figure(figsize=(10, 5))
andrews_curves(df, "Species", color=["red", "blue", "green"], alpha=0.3)
plt.title("Andrews Curves")
plt.ylabel("f(t)")
plt.tight_layout()
plt.savefig("lab03_11_andrews.png", dpi=100)
plt.close()

# Findings: Andrews curves for Setosa form a visually distinct band
# from the other two species across the full t range, reflecting their
# separability. Versicolor and Virginica curves partially overlap,
# consistent with their closer feature distributions.

```

```python
# ─────────────────────────────────────────────
# 12. Correlation Heatmap
# ─────────────────────────────────────────────
plt.figure(figsize=(7, 5))
sns.heatmap(df[features].corr(), annot=True, fmt=".2f", cmap="coolwarm", square=True)
plt.title("Correlation Heatmap of Iris Features")
plt.tight_layout()
plt.savefig("lab03_12_correlation_heatmap.png", dpi=100)
plt.close()

# Findings: PetalLengthCm and PetalWidthCm are strongly positively
# correlated (r ≈ 0.96). SepalLengthCm also correlates strongly with
# both petal features. SepalWidthCm shows weak or slightly negative
# correlation with the rest.
print("All plots saved.")
```
