```python
import numpy as np
from sklearn.linear_model import LinearRegression

car_age = [5,7,8,7,2,17,2,9,4,11,12,9,6]
car_speed = [99,86,87,88,111,86,103,87,94,78,77,85,86]

X = np.array(car_age).reshape(-1, 1)
y = np.array(car_speed)

model = LinearRegression()
model.fit(X, y)
```

```python
for i in car_age:
    speed=model.predict([[i]])[0]
    print(f"{i} years - {speed:.2f} km/h")
```

```python

age_input = float(input("Enter car age: "))
predicted_speed = model.predict([[age_input]])[0]
print(f"Predicted speed for age {age_input}: {predicted_speed:.2f}")
```

```python

import matplotlib.pyplot as plt

x_line = np.linspace(min(car_age), max(car_age), 100).reshape(-1, 1)
y_line = model.predict(x_line)

plt.scatter(car_age, car_speed, color='blue', label='Data points')
plt.plot(x_line, y_line, color='red', label='Regression line')
plt.scatter([age_input], [predicted_speed], color='green', zorder=5, label=f'Predicted: age={age_input}, speed={predicted_speed:.2f}')
plt.xlabel('Car Age')
plt.ylabel('Car Speed')
plt.title('Car Age vs Speed - Linear Regression')
plt.legend()
plt.savefig('../Outputs/Lab-10_Output/Lab10_Car.png')
plt.show()
print(f"Intercept: {model.intercept_:.4f}, Coefficient: {model.coef_[0]:.4f}")
```

```python
age=[18,22,30,45,65,80]
accident_no=[38,36,24,20,18,28]

X=np.array(age).reshape(-1,1)
y=np.array(accident_no)

Model=LinearRegression()
Model.fit(X,y)
```

```python
for i in age:
    accidents=Model.predict([[i]])[0]
    print(f"{accidents:.2f} accidents - {i} years old")
```

```python
in_age=float(input("Enter age : "))
accident=Model.predict([[in_age]])[0]
print(f"{accident:.2f} accidents for age {in_age}")
```

```python

import matplotlib.pyplot as plt

x_line = np.linspace(min(age), max(age), 100).reshape(-1, 1)
y_line = Model.predict(x_line)

plt.scatter(age, accident_no, color='blue', label='Data points')
plt.plot(x_line, y_line, color='red', label='Regression line')
plt.scatter([in_age], [accident], color='green', zorder=5, label=f'Predicted: age={in_age}, speed={accident:.2f}')
plt.xlabel('Age')
plt.ylabel('Accidents')
plt.title('Age vs Accidents - Linear Regression')
plt.legend()
plt.savefig('../Outputs/Lab-10_Output/Lab10_Accident.png')
plt.show()
print(f"Intercept: {model.intercept_:.4f}, Coefficient: {model.coef_[0]:.4f}")
```
