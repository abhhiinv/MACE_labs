```python
import pandas as pd
import numpy as np
import matplotlib as plt
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
```

```python
df=pd.read_csv('../Datasets/Wine_Quality.csv')
print(df.columns)
print(df.isnull().sum())
print(df.describe())
df.dropna(inplace=True)
print(df.isnull().sum())
df.describe()
```

```python
print(df.dtypes)
le=LabelEncoder()
df['type']=le.fit_transform(df['type'])
ratios = [(0.1,"90-10"),(0.2,"80-20"),(0.3,"70-30"),(0.4,"60-40")]
print(df.dtypes)
```

```python
X=df.drop('quality',axis=1).values
y=df['quality'].values
for ratio,text in ratios:
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=ratio, random_state=42)
    gnb = GaussianNB()
    gnb.fit(X_train, y_train)
    predicted = gnb.predict(X_test)
    print(f"Classification report for {text} split\n {classification_report(y_test, predicted,zero_division=0)}")
    print(f"Confusion matrix for {text} split\n {confusion_matrix(y_test, predicted)}")
    print(f"Accuracy for {text} split = {accuracy_score(y_test, predicted)}\n\n")
    print("_"*100)
```
