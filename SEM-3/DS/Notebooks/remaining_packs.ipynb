**POLARS**

Polars is a fast DataFrame library written in Rust and designed for performance, offering a syntax similar to Pandas but optimized for large datasets and multi-core processing. It supports lazy evaluation, meaning operations are only executed when explicitly triggered, which allows the engine to optimize the query plan before running.


1. Creating a DataFrame

Create a table from a dictionary.

```python
import polars as pl
df = pl.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [25, 30, 35],
    'salary': [50000, 60000, 75000]
})
print(df)
```

2. Selecting and filtering

Select columns and filter rows by condition.

```python
print(df.select(['name', 'salary']))
print(df.filter(pl.col('age') > 25))
```

3. Adding a new column

Derive a new column from existing ones.

```python
df = df.with_columns(
    (pl.col('salary') * 0.1).alias('bonus')
)
print(df)
```

4. GroupBy and aggregation

Group rows and compute aggregate statistics.

```python
df = pl.DataFrame({
    'dept': ['HR', 'IT', 'HR', 'IT', 'Finance'],
    'salary': [45000, 72000, 51000, 68000, 56000]
})
result = df.group_by('dept').agg(pl.col('salary').mean().alias('avg_salary'))
print(result)
```

5. Lazy evaluation

Build a query plan and execute it only when needed.

```python
result = (
    pl.DataFrame({'a': [1, 2, 3, 4, 5], 'b': [10, 20, 30, 40, 50]})
    .lazy()
    .filter(pl.col('a') > 2)
    .with_columns((pl.col('b') * 2).alias('b_doubled'))
    .collect()
)
print(result)
```

**HUGGING FACE TRANSFORMERS**

Hugging Face Transformers is a Python library providing thousands of pre-trained models for natural language processing, computer vision, and audio tasks. It allows you to use state-of-the-art models like BERT, GPT-2, and T5 with just a few lines of code through its high-level pipeline API.

1. Sentiment analysis

Classify text as positive or negative.

```python
from transformers import pipeline
classifier = pipeline('sentiment-analysis')
result = classifier('I love learning data science!')
print(result)
```

2. Text generation

Generate text continuation from a prompt.

```python
from transformers import pipeline
generator = pipeline('text-generation', model='gpt2')
result = generator('Python is a great language for', max_length=30, num_return_sequences=1)
print(result[0]['generated_text'])
```

3. Named entity recognition

Identify and classify named entities in text.

```python
from transformers import pipeline
ner = pipeline('ner', grouped_entities=True)
result = ner('Sundar Pichai is the CEO of Google based in California.')
for entity in result:
    print(entity)
```

4. Text summarization

Summarize a long piece of text.

```python
from transformers import pipeline
summarizer = pipeline('summarization')
text = """Data science is an interdisciplinary field that uses scientific methods,
processes, algorithms, and systems to extract knowledge and insights from structured
and unstructured data. It combines statistics, mathematics, and computer science."""
summary = summarizer(text, max_length=50, min_length=20)
print(summary[0]['summary_text'])
```

5. Question answering

Extract an answer from a context passage given a question.

```python
from transformers import pipeline
qa = pipeline('question-answering')
result = qa(
    question='What is Python used for?',
    context='Python is a programming language widely used for data science, web development, and automation.'
)
print(result['answer'])
```

**PLOTLY**

Plotly is a Python library for creating interactive, publication-quality charts and graphs that render in web browsers. Unlike Matplotlib, Plotly charts support zooming, panning, hover tooltips, and other interactive features out of the box.

1. Line chart

Create an interactive line plot.

```python
import plotly.express as px
df = px.data.gapminder().query("country == 'India'")
fig = px.line(df, x='year', y='gdpPercap', title='India GDP per Capita')
fig.show()
```

2. Bar chart

Create an interactive bar chart.

```python
import plotly.express as px
df = px.data.tips()
fig = px.bar(df, x='day', y='total_bill', color='sex', title='Total Bill by Day')
fig.show()
```

3. Scatter plot

Create an interactive scatter plot with color grouping.

```python
import plotly.express as px
df = px.data.iris()
fig = px.scatter(df, x='sepal_width', y='sepal_length',
                 color='species', title='Iris Sepal Dimensions')
fig.show()
```

4. Pie chart

Show proportions as an interactive pie chart.

```python
import plotly.express as px
df = px.data.tips()
fig = px.pie(df, names='day', values='total_bill', title='Bill Share by Day')
fig.show()
```

5. 3D scatter plot

Visualize three dimensions simultaneously.

```python
import plotly.express as px
df = px.data.iris()
fig = px.scatter_3d(df, x='sepal_length', y='sepal_width',
                    z='petal_length', color='species')
fig.show()
```

**REQUESTS**

Requests is a Python library for making HTTP requests in a simple and human-friendly way, supporting GET, POST, PUT, DELETE, and other methods. It is widely used for consuming REST APIs, downloading web content, and interacting with web services.

1. GET request

Fetch data from a public API.

```python
import requests
response = requests.get('https://jsonplaceholder.typicode.com/posts/1')
print('Status:', response.status_code)
print('Data:', response.json())
```

2. GET with parameters

Pass query parameters with a request.

```python
import requests
params = {'userId': 1}
response = requests.get('https://jsonplaceholder.typicode.com/posts', params=params)
data = response.json()
print(f'Posts fetched: {len(data)}')
print(data[0])
```

3. POST request

Send data to a server.

```python
import requests
payload = {'title': 'My Post', 'body': 'Hello world', 'userId': 1}
response = requests.post('https://jsonplaceholder.typicode.com/posts', json=payload)
print('Status:', response.status_code)
print('Response:', response.json())
```

4. Response headers and status

Inspect response metadata.

```python
import requests
response = requests.get('https://jsonplaceholder.typicode.com/users/1')
print('Status code:', response.status_code)
print('Content type:', response.headers['Content-Type'])
user = response.json()
print('User name:', user['name'])
```

5. Error handling

Handle failed requests gracefully.

```python
import requests
try:
    response = requests.get('https://jsonplaceholder.typicode.com/posts/999', timeout=5)
    response.raise_for_status()
    print(response.json())
except requests.exceptions.HTTPError as e:
    print('HTTP error:', e)
except requests.exceptions.ConnectionError:
    print('Connection failed')
except requests.exceptions.Timeout:
    print('Request timed out')
```
