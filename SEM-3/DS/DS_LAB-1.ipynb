## SCIKIT-LEARN
Scikit-learn is a machine learning library that provides tools for classification, regression, clustering, and dimensionality reduction. It includes pre-built algorithms like decision trees, support vector machines, and neural networks along with utilities for model evaluation and data preprocessing.

### 1. Train-test split
Divide data for training and testing.

```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
print(X_train.shape, X_test.shape)
```

### 2. Decision Tree classifier
Train classifier and make prediction.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)
clf = DecisionTreeClassifier()
clf.fit(X, y)
pred = clf.predict([[5.1, 3.5, 1.4, 0.2]])
print(pred)
```

### 3. Scaling features
Normalize features to 0 mean and unit variance.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled[:2])
```

### 4. Model accuracy
Evaluate model performance on test data.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
clf = DecisionTreeClassifier()
clf.fit(X_train, y_train)
accuracy = clf.score(X_test, y_test)
print(f"Accuracy: {accuracy}")
```

### 5. Cross-validation
Test model across multiple data splits.

```python
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)
clf = DecisionTreeClassifier()
scores = cross_val_score(clf, X, y, cv=5)
print(f"CV scores: {scores}, Mean: {scores.mean()}")
```

## KERAS
Keras is a deep learning library used to build and train neural networks with simple, user-friendly APIs. It allows you to define layers, compile models, and train them on data for tasks like image recognition, natural language processing, and time series forecasting.

### 1. Sequential model
Create simple neural network architecture.

```python
import keras
from keras.models import Sequential
from keras.layers import Dense
model = Sequential([
    Dense(10, activation='relu', input_dim=4),
    Dense(3, activation='softmax')
])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
print(model.summary())
```

### 2. Training model
Train network on data.

```python
import keras
from keras.models import Sequential
from keras.layers import Dense
from keras.datasets import mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()
X_train = X_train.reshape(-1, 784).astype('float32') / 255
X_test = X_test.reshape(-1, 784).astype('float32') / 255
model = Sequential([Dense(128, activation='relu', input_dim=784), Dense(10, activation='softmax')])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=5, batch_size=32)
```

### 3. Model prediction
Predict on new data.

```python
predictions = model.predict(X_test[:5])
print(predictions)
```

### 4. Layers and activation
Build layer-by-layer, showing different activations.

```python
from keras.models import Sequential
from keras.layers import Dense
model = Sequential()
model.add(Dense(64, activation='relu', input_dim=10))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
print(model.summary())
```

### 5. Evaluation
Test trained model on test set.

```python
import numpy as np
from keras.models import Sequential
from keras.layers import Dense

# Generate dummy data matching input_dim=10
X_dummy = np.random.rand(100, 10).astype("float32")
y_dummy = np.random.randint(0, 2, 100).astype("float32")

model = Sequential()
model.add(Dense(64, activation="relu", input_dim=10))
model.add(Dense(32, activation="relu"))
model.add(Dense(1, activation="sigmoid"))
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
model.fit(X_dummy, y_dummy, epochs=3, batch_size=16, verbose=0)

loss, accuracy = model.evaluate(X_dummy, y_dummy)
print(f"Loss: {loss}, Accuracy: {accuracy}")
```

## SCIPY
SciPy is a Python library built on NumPy that provides algorithms for scientific and technical computing including optimization, integration, interpolation, eigenvalue problems, and statistics. It is widely used in fields like physics, engineering, and data science for numerical problem solving.

### 1. Statistical functions
Compute mean, standard deviation, and t-test.

```python
from scipy import stats
import numpy as np
data = np.array([12, 15, 14, 10, 18, 22, 13])
print('Mean:', np.mean(data))
print('Std:', np.std(data))
t_stat, p_val = stats.ttest_1samp(data, 15)
print(f't-statistic: {t_stat:.4f}, p-value: {p_val:.4f}')
```

### 2. Linear algebra
Solve a system of linear equations.

```python
from scipy.linalg import solve
import numpy as np
# Solve: 2x + y = 5, x + 3y = 10
A = np.array([[2, 1], [1, 3]])
b = np.array([5, 10])
x = solve(A, b)
print('Solution x, y:', x)
```

### 3. Numerical integration
Compute definite integral of a function.

```python
from scipy.integrate import quad
import numpy as np
# Integrate sin(x) from 0 to pi
result, error = quad(np.sin, 0, np.pi)
print(f'Integral of sin(x) from 0 to pi: {result:.4f}')
```

### 4. Optimization
Find the minimum of a function.

```python
from scipy.optimize import minimize
def func(x):
    return (x - 3) ** 2 + 2
result = minimize(func, x0=0)
print('Minimum at x =', result.x)
print('Minimum value =', result.fun)
```

### 5. Training loop
Manually train a model using loss and optimizer.

```python
import torch
import torch.nn as nn
X = torch.rand(100, 4)
y = torch.randint(0, 2, (100, 1)).float()
model = nn.Sequential(nn.Linear(4, 8), nn.ReLU(), nn.Linear(8, 1), nn.Sigmoid())
criterion = nn.BCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
for epoch in range(5):
    pred = model(X)
    loss = criterion(pred, y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    print(f'Epoch {epoch+1}, Loss: {loss.item():.4f}')
```

## NLTK
NLTK (Natural Language Toolkit) is a Python library for working with human language data, providing tools for text processing tasks like tokenization, stemming, tagging, and parsing. It is widely used in natural language processing and computational linguistics for text analysis and building language models.

### 1. Tokenization
Split text into words and sentences.

```python
import nltk
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
text = 'Natural language processing is fun. NLTK makes it easy.'
words = nltk.word_tokenize(text)
sentences = nltk.sent_tokenize(text)
print('Words:', words)
print('Sentences:', sentences)
```

### 2. Stopword removal
Filter out common words with no meaningful content.

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords', quiet=True)
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
text = 'This is a simple example of stopword removal in NLTK'
words = nltk.word_tokenize(text)
stop_words = set(stopwords.words('english'))
filtered = [w for w in words if w.lower() not in stop_words]
print('Filtered:', filtered)
```

### 3. Stemming
Reduce words to their root form.

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
words = ['running', 'flies', 'happily', 'studies', 'played']
stemmed = [stemmer.stem(w) for w in words]
print('Stemmed:', stemmed)
```

### 4. POS tagging
Label each word with its part of speech.

```python
import nltk
nltk.download('averaged_perceptron_tagger', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
text = 'Python is a powerful programming language'
tokens = nltk.word_tokenize(text)
tags = nltk.pos_tag(tokens)
print(tags)
```

## PYBRAIN
PyBrain is a Python machine learning library focused on neural networks, reinforcement learning, and evolutionary algorithms with a simple and modular design. Though no longer actively maintained, it is used in academic settings to understand fundamental neural network building blocks.

### Note
PyBrain may not be available in all environments. Install with: pip install pybrain

### 1. Build a feedforward network
Create a simple neural network with input, hidden, and output layers.

```python
from pybrain.structure import FeedForwardNetwork, LinearLayer, SigmoidLayer, FullConnection
net = FeedForwardNetwork()
inLayer = LinearLayer(2)
hiddenLayer = SigmoidLayer(3)
outLayer = LinearLayer(1)
net.addInputModule(inLayer)
net.addModule(hiddenLayer)
net.addOutputModule(outLayer)
net.addConnection(FullConnection(inLayer, hiddenLayer))
net.addConnection(FullConnection(hiddenLayer, outLayer))
net.sortModules()
print(net)
```

### 2. Activate network
Pass input through the network to get output.

```python
from pybrain.structure import FeedForwardNetwork, LinearLayer, SigmoidLayer, FullConnection
net = FeedForwardNetwork()
inLayer = LinearLayer(2)
hiddenLayer = SigmoidLayer(3)
outLayer = LinearLayer(1)
net.addInputModule(inLayer)
net.addModule(hiddenLayer)
net.addOutputModule(outLayer)
net.addConnection(FullConnection(inLayer, hiddenLayer))
net.addConnection(FullConnection(hiddenLayer, outLayer))
net.sortModules()
output = net.activate([1.0, 0.5])
print('Output:', output)
```

### 1. Basic spider
Define a spider that crawls a website.

```python
import scrapy

class QuotesSpider(scrapy.Spider):
    name = 'quotes'
    start_urls = ['http://quotes.toscrape.com']

    def parse(self, response):
        for quote in response.css('div.quote'):
            yield {
                'text': quote.css('span.text::text').get(),
                'author': quote.css('small.author::text').get(),
            }
```

### 2. CSS selectors
Extract data from HTML using CSS selectors.

```python
def parse(self, response):
    title = response.css('h1::text').get()
    link = response.css('a::attr(href)').get()
    all_links = response.css('a::attr(href)').getall()
    print(title, link, all_links)
```

### 3. Following links
Crawl multiple pages by following links.

```python

def parse(self, response):
    for quote in response.css('div.quote'):
        yield {'text': quote.css('span.text::text').get()}

    next_page = response.css('li.next a::attr(href)').get()
    if next_page:
        yield response.follow(next_page, self.parse)
```

## BIOPYTHON
BioPython is a Python library for bioinformatics that provides tools for working with biological sequence data, parsing bioinformatics file formats, and accessing biological databases. ### 1. Create a DNA sequence
Create and analyze a biological sequence.

```python
from Bio.Seq import Seq
dna = Seq('ATGGCCATTGTAATGGGCCGCTGA')
print('DNA:', dna)
print('Length:', len(dna))
print('GC count:', dna.count('G') + dna.count('C'))
```

### 2. Complement and reverse complement
Get complementary DNA strands.

```python
from Bio.Seq import Seq
dna = Seq('ATGGCCATT')
print('Original:', dna)
print('Complement:', dna.complement())
print('Reverse complement:', dna.reverse_complement())
```

### 3. Transcription and translation
Convert DNA to RNA to protein.

```python
from Bio.Seq import Seq
dna = Seq('ATGGCCATTGTAATGGGCCGCTGA')
rna = dna.transcribe()
protein = dna.translate()
print('DNA:', dna)
print('RNA:', rna)
print('Protein:', protein)
```

## PYGAME
PyGame is a Python library for creating 2D games and multimedia applications by providing functionality for graphics, sound, keyboard/mouse input, and game loop management. It is built on top of SDL and is commonly used in game development education and hobby projects.

### Note
PyGame requires a display environment to render graphics and cannot run inside Jupyter notebooks. The examples below demonstrate the structure and patterns used in PyGame programs.

### 1. Basic game window
Initialize PyGame and create a window.

```python

import pygame
pygame.init()
screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption('My Game')
screen.fill((0, 0, 0))
pygame.display.flip()
# pygame.quit()
```

### 2. Game loop structure
Core structure of any PyGame application.

```python
import pygame
pygame.init()
screen = pygame.display.set_mode((800, 600))
clock = pygame.time.Clock()
running = True

while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    screen.fill((30, 30, 30))
    pygame.display.flip()
    clock.tick(60)

# pygame.quit()
```

### 3. Drawing shapes
Draw rectangles, circles, and lines on screen.

```python

import pygame
pygame.init()
screen = pygame.display.set_mode((400, 400))
screen.fill((255, 255, 255))

pygame.draw.rect(screen, (255, 0, 0), (50, 50, 100, 80))
pygame.draw.circle(screen, (0, 0, 255), (200, 200), 60)
pygame.draw.line(screen, (0, 255, 0), (0, 0), (400, 400), 3)

pygame.display.flip()
# pygame.quit()
```

### 4. Keyboard input
Detect keyboard events inside the game loop.

```python
import pygame
pygame.init()
screen = pygame.display.set_mode((400, 400))
running = True

while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
        if event.type == pygame.KEYDOWN:
            if event.key == pygame.K_LEFT:
                print('Left arrow pressed')
            if event.key == pygame.K_RIGHT:
                print('Right arrow pressed')

# pygame.quit()
```
