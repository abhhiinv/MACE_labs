## Import Pandas

```python
import pandas as pd
import numpy as np
```

## Sample Data
Load the Iris dataset from scikit-learn and convert to a Pandas DataFrame.

```python
from sklearn.datasets import load_iris
iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target  # 0=setosa, 1=versicolor, 2=virginica
# print(df.head())
```

## 1. DataFrame()
Creates a 2D tabular data structure with rows and columns, similar to a spreadsheet or SQL table.

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target
print(df)
```

## 2. Series()
Creates a 1D labeled array capable of holding any data type. It is essentially a single column of a DataFrame.

```python
sepal_length = pd.Series(df['sepal length (cm)'].values, name='sepal length')
print(sepal_length.head(10))
```

## 3. info()
Prints a concise summary of the DataFrame including column names, data types, non-null counts, and memory usage.

```python
df.info()
```

## 4. shape
Returns a tuple representing the dimensions of the DataFrame as (number of rows, number of columns).

```python
print('Shape:', df.shape)
print('Rows:', df.shape[0])
print('Columns:', df.shape[1])
```

## 5. describe()
Generates descriptive statistics for all numeric columns including count, mean, std, min, max, and quartile values.

```python
print(df.describe())
```

## 6. head()
Returns the first n rows of the DataFrame. Default is 5 rows.

```python
print(df.head())
print()
print('First 3 rows:')
print(df.head(3))
```

## 7. tail()
Returns the last n rows of the DataFrame. Default is 5 rows.

```python
print(df.tail())
print()
print('Last 2 rows:')
print(df.tail(2))
```

## 8. count()
Returns the number of non-null values in each column. Useful for detecting missing data.

```python
print(df.count())
```

## 9. sum()
Returns the sum of values for each numeric column.

```python
print(df[iris.feature_names].sum())
```

## 10. mean()
Returns the arithmetic mean of values for each numeric column.

```python
print(df[iris.feature_names].mean())
```

## 11. median()
Returns the median (middle value) of each numeric column. Less sensitive to outliers than mean.

```python
print(df[iris.feature_names].median())
```

## 12. mode()
Returns the most frequently occurring value(s) in each column.

```python
print(df[['sepal length (cm)', 'petal length (cm)', 'species']].mode())
```

## 13. std()
Returns the standard deviation of values for each numeric column, measuring the spread of data.

```python
print(df[iris.feature_names].std())
```

## 14. min()
Returns the minimum value in each numeric column.

```python
print(df[iris.feature_names].min())
```

## 15. max()
Returns the maximum value in each numeric column.

```python
print(df[iris.feature_names].max())
```

## 16. abs()
Returns the absolute value of each element. Useful when working with negative numbers.

```python
diff = df['sepal length (cm)'] - df['sepal length (cm)'].mean()
print('Deviations (first 5):')
print(diff.head())
print('Absolute values (first 5):')
print(diff.abs().head())
```

## 17. prod()
Returns the product of all values in each numeric column.

```python
small = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
print('DataFrame:')
print(small)
print('Product:')
print(small.prod())
```

## 18. cumsum()
Returns the cumulative sum of values along a column. Each row shows the running total up to that point.

```python
print('Sepal length values (first 6):')
print(df['sepal length (cm)'].head(6).values)
print('Cumulative Sum:')
print(df['sepal length (cm)'].cumsum().head(6).values)
```

## 19. cumprod()
Returns the cumulative product of values along a column. Each row shows the running product up to that point.

```python
small_series = pd.Series([1, 2, 3, 4, 5])
print('Values:', small_series.values)
print('Cumulative Product:', small_series.cumprod().values)
```

## 20. read_csv()
Loads data from a CSV (Comma Separated Values) file into a DataFrame.

```python
df.to_csv('iris.csv', index=False)

df_csv = pd.read_csv('iris.csv')
print(df_csv.head())
```

## 21. read_excel()
Loads data from an Excel (.xlsx or .xls) file into a DataFrame.

```python
df.to_excel('iris.xlsx', index=False)

df_excel = pd.read_excel('iris.xlsx')
print(df_excel.head())
```
